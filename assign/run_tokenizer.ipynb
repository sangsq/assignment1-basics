```python
from tokenization import *
import numpy as np
import torch
```

## Build tokenizer

```python
input_path = "../data/TinyStoriesV2-GPT4-train.txt"
name = "tiny-stories-v2-gpt4-train"
vocab_size = 10000
special_tokens = ["<|endoftext|>"]
```

```python
PAT = r"""'(?:[sdmt]|ll|ve|re)| ?\p{L}+| ?\p{N}+| ?[^\s\p{L}\p{N}]+|\s+(?!\S)|\s+"""
with open(input_path, "r", encoding="utf-8") as f:
    text = f.read()
pre_tokens = pre_tokenization(text, PAT, special_tokens)
vocab, merges = construct_bpe(pre_tokens, vocab_size, special_tokens)
```

```python
tokenizer = Tokenizer(vocab, merges, special_tokens)
torch.save(tokenizer, f"../tokenizers/{name}_tokenizer.pt")
```

## Encoding text

```python
text_path = "../data/TinyStoriesV2-GPT4-train.txt"
output_path = "../data/tiny-stories-v2-gpt4-train_encoded.npy"
tokenizer_path = "../tokenizers/tiny-stories-v2-gpt4-train_tokenizer.pt"


tokenizer = torch.load(tokenizer_path, weights_only=False)
with open(text_path, "r", encoding="utf-8") as f:
    text = f.read()
encoded_text = tokenizer.encode(text)
tmp = np.array(encoded_text, dtype=np.int16)
np.save(output_path, tmp)
```
